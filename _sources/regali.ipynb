```python
# Cell 1: Imports
import ipywidgets as widgets
from IPython.display import display
from cryptography.fernet import Fernet

# -----------------------------
# Load master key
# -----------------------------
with open("master_key.txt", "rb") as f:
    master_key = f.read().strip()

cipher = Fernet(master_key)

# -----------------------------
# Load encrypted assignments
# -----------------------------
assignments = {}
with open("assignments.txt", "r") as f:
    for line in f.read().splitlines():
        giver, encrypted, short_key = line.split(",")
        assignments[short_key] = encrypted

# -----------------------------
# Define decryption function
# -----------------------------
def reveal_giftee(short_key):
    if short_key not in assignments:
        output.value = f"<b style='color:red'>Invalid key: {short_key}</b>"
        return
    encrypted_payload = assignments[short_key]
    decrypted = cipher.decrypt(encrypted_payload.encode()).decode()
    key_in_file, receiver = decrypted.split(":")
    if key_in_file != short_key:
        output.value = "<b style='color:red'>Key mismatch — possible tampering.</b>"
    else:
        output.value = f"<b style='color:green'>You give a gift to: {receiver}</b>"

# -----------------------------
# Create widget
# -----------------------------
text = widgets.Text(
    value='',
    placeholder='Enter your key',
    description='Key:',
    disabled=False
)
button = widgets.Button(description="Reveal Giftee")
output = widgets.HTML(value="")

def on_button_clicked(b):
    reveal_giftee(text.value.strip())

button.on_click(on_button_clicked)

display(text, button, output)
```
